## 5. Preprocessing and transformations
<hr>
rubric={points:10}

<div class="alert alert-info">
    
**Your tasks:**

1. **Identify feature types**: Determine the different types of features in your dataset (e.g., numerical, categorical, ordinal, text) and specify the transformations you would apply to each type.

2. **Define a column transformer (if needed)**: Implement a `ColumnTransformer` to apply the appropriate preprocessing steps to each feature type.

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
````

Numerical features include latitude, longitude, minimum_nights, number_of_reviews, reviews_per_month, calculated_host_listings_count, and availability_365.
Categorical features include neighbourhood_group, neighbourhood, and room_type.
ID and text-based columns are removed as they do not provide predictive value.

````python
numeric_features = [
    "latitude",
    "longitude",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365"
]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])
````

````python
categorical_features = [
    "neighbourhood_group",
    "neighbourhood",
    "room_type"
]

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
````

````python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
````

````python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
````

Numerical features were imputed using the mean and standardized to ensure comparable scales.
Categorical variables were imputed using the most frequent value and encoded using OneHotEncoder.
A ColumnTransformer was used to apply appropriate preprocessing steps to each feature type.

````python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
import numpy as np
linear_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

linear_model.fit(X_train, y_train)

y_pred_linear = linear_model.predict(X_test)

rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))

print("Linear Regression RMSE:", rmse_linear)
````

````python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV

ridge_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", Ridge())
])

param_grid = {
    "model__alpha": [0.1, 1.0, 10.0, 100.0]
}

grid_search = GridSearchCV(
    ridge_pipeline,
    param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error"
)

grid_search.fit(X_train, y_train)

print("Best alpha:", grid_search.best_params_)
print("Best CV RMSE:", -grid_search.best_score_)
````

````python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_validate

from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor


models = {
    "Decision Tree": DecisionTreeRegressor(random_state=123, max_depth=10),
    "Random Forest": RandomForestRegressor(
        random_state=123,
        n_estimators=200,
        max_depth=None,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(random_state=123),
    "KNN": KNeighborsRegressor(n_neighbors=10)
}

results = []

for name, model in models.items():
    pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    cv_out = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=5,
        scoring="neg_root_mean_squared_error",
        return_train_score=True
    )

    train_rmse = -cv_out["train_score"]
    val_rmse = -cv_out["test_score"]

    results.append({
        "Model": name,
        "Train RMSE (mean)": train_rmse.mean(),
        "Train RMSE (std)": train_rmse.std(),
        "CV RMSE (mean)": val_rmse.mean(),
        "CV RMSE (std)": val_rmse.std(),
        "Fit time (mean)": cv_out["fit_time"].mean(),
        "Score time (mean)": cv_out["score_time"].mean()
    })

results_df = pd.DataFrame(results).sort_values("CV RMSE (mean)")
results_df
````

None of the non-linear models outperformed the linear model. Decision Tree and Random Forest show clear overfitting, while Gradient Boosting and KNN are more stable but do not improve performance. Tree ensembles require longer training times. Overall, the linear model remains the most effective approach for this dataset.

````python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
import numpy as np

gbr_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", GradientBoostingRegressor(random_state=123))
])

gbr_param_grid = {
    "model__n_estimators": [100, 300],
    "model__learning_rate": [0.05, 0.1],
    "model__max_depth": [2, 3],
    "model__subsample": [0.8, 1.0]
}

gbr_grid = GridSearchCV(
    gbr_pipe,
    param_grid=gbr_param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

gbr_grid.fit(X_train, y_train)

print("Best GBR params:", gbr_grid.best_params_)
print("Best GBR CV RMSE:", -gbr_grid.best_score_)
````

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(random_state=123, n_jobs=-1))
])

rf_param_dist = {
    "model__n_estimators": [100, 200, 400],
    "model__max_depth": [None, 10, 20, 30],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt", 0.5, 1.0]
}

rf_random = RandomizedSearchCV(
    rf_pipe,
    param_distributions=rf_param_dist,
    n_iter=12,                 # 控制时间：12次随机搜索
    cv=5,
    scoring="neg_root_mean_squared_error",
    random_state=123,
    n_jobs=-1
)

rf_random.fit(X_train, y_train)

print("Best RF params:", rf_random.best_params_)
print("Best RF CV RMSE:", -rf_random.best_score_)
````

````python
import pandas as pd

summary = pd.DataFrame([
    {"Model": "Linear/Ridge (previous)", "Best CV RMSE": 222.82},
    {"Model": "Gradient Boosting (tuned)", "Best CV RMSE": -gbr_grid.best_score_},
    {"Model": "Random Forest (tuned)", "Best CV RMSE": -rf_random.best_score_}
])

summary
````

Hyperparameter optimization was performed using GridSearchCV for Gradient Boosting and RandomizedSearchCV for Random Forest, tuning multiple hyperparameters for each model. The tuned Gradient Boosting model achieved a CV RMSE of approximately 219.83, while the tuned Random Forest achieved the best performance with a CV RMSE of 216.00. Compared to the linear/Ridge model (CV RMSE ≈ 222.82), hyperparameter optimization led to meaningful performance improvements. These results indicate that properly tuned tree-based ensemble models can outperform linear models on this dataset.
